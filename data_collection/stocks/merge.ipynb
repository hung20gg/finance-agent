```python
import pandas as pd
```

```python
OLD_STOCK_DIR = '../data/stock_daily.parquet'
NEW_STOCK_DIR = '../data/vn_stock_price_daily_yahoo.parquet'
```

```python
df_old = pd.read_parquet(OLD_STOCK_DIR)
df_new = pd.read_parquet(NEW_STOCK_DIR)
```

```python
df_old.head()
```

```python
df_old[(df_old['stock_code'] == 'VIC') & (df_old['time'] > '2016-01-01')].sort_values('time').head()
```

```python
df_new[df_new['stock_code'] == 'VIC'].head()
```

```python
df_new['close'] = df_new['close'] / 1000
df_new['EMA20'] = df_new['EMA20'] / 1000
df_new['EMA50'] = df_new['EMA50'] / 1000
df_new['EMA200'] = df_new['EMA200'] / 1000
```

```python
df_old_stock_code = set(df_old['stock_code'].unique())
df_new_stock_code = set(df_new['stock_code'].unique())
```

```python
df_old_stock_code
```

```python
len(df_old_stock_code)
```

```python
diff_stock_code = df_new_stock_code - df_old_stock_code
diff_stock_code
```

```python
df_new_new = df_new[df_new['stock_code'].isin(diff_stock_code)]
df_new_new.head()
```

```python
df_new_new.drop(columns=['yahoo_symbol'], inplace=True)
```

```python
df_new_new.to_parquet('../data/stock_price_daily_yahoo.parquet', index=False)
```
